### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento



```python
faturamento_loja = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

```

```python
print(f"R$ {faturamento_loja:,.2f}")
print(f"R$ {faturamento_loja2:,.2f}")
print(f"R$ {faturamento_loja3:,.2f}")
print(f"R$ {faturamento_loja4:,.2f}")
```

# 2. Vendas por Categoria

```python
categorias_loja1 = loja.groupby('Categoria do Produto').size()
categorias_loja2 = loja2.groupby('Categoria do Produto').size()
categorias_loja3 = loja3.groupby('Categoria do Produto').size()
categorias_loja4 = loja4.groupby('Categoria do Produto').size()
```

```python
print('Categorias Loja 1:', categorias_loja1)
print('Categorias Loja 2:', categorias_loja2)
```

```python
print('Categorias Loja 3:', categorias_loja3)   
print('Categorias Loja 4:', categorias_loja4)
```

# 3. Média de Avaliação das Lojas


```python

media_loja1 = loja['Avaliação da compra'].mean()
media_loja2 = loja2['Avaliação da compra'].mean()
media_loja3 = loja3['Avaliação da compra'].mean()
media_loja4 = loja4['Avaliação da compra'].mean()
```

```python

print(f"Média Loja 1: {media_loja1:.2f}")
print(f"Média Loja 2: {media_loja2:.2f}")
print(f"Média Loja 3: {media_loja3:.2f}")
print(f"Média Loja 4: {media_loja4:.2f}")
```

# 4. Produtos Mais e Menos Vendidos

```python
frequencia_produtos_loja1 = loja['Produto'].value_counts()
frequencia_produtos_loja2 = loja2['Produto'].value_counts()
frequencia_produtos_loja3 = loja3['Produto'].value_counts()
frequencia_produtos_loja4 = loja4['Produto'].value_counts()
```

```python
top_tres_loja1 = frequencia_produtos_loja1.head(3)
bottom_tres_loja1 = frequencia_produtos_loja1.tail(3)
top_tres_loja2 = frequencia_produtos_loja2.head(3)
bottom_tres_loja2 = frequencia_produtos_loja2.tail(3)
top_tres_loja3 = frequencia_produtos_loja3.head(3)
bottom_tres_loja3 = frequencia_produtos_loja3.tail(3)
top_tres_loja4 = frequencia_produtos_loja4.head(3)
bottom_tres_loja4 = frequencia_produtos_loja4.tail(3)
```

```python
print("Loja 1 - Top 3 Produtos:\n", top_tres_loja1)
print("Loja 1 - Bottom 3 Produtos:\n", bottom_tres_loja1)
print()
print("Loja 2 - Top 3 Produtos:\n", top_tres_loja2)
print("Loja 2 - Bottom 3 Produtos:\n", bottom_tres_loja2)
print()
print("Loja 3 - Top 3 Produtos:\n", top_tres_loja3)
print("Loja 3 - Bottom 3 Produtos:\n", bottom_tres_loja3)
print()
print("Loja 4 - Top 3 Produtos:\n", top_tres_loja4)
print("Loja 4 - Bottom 3 Produtos:\n", bottom_tres_loja4)
```
